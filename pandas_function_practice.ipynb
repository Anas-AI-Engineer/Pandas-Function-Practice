```python
import pandas as pd 

data = {
    "Name": ["Ali", "Ahmed", "Anas", "Sara"],
    "Age": [20, 30, 40, 50],
    "Marks": [50, 45, 35, 87],
    "City": ["Karcahi", "Lahore", "Islamabad", "Quetta"]
}

df = pd.DataFrame(data)

print(data)
print(df.head())
print(df.head(2))
print(df.tail())
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
print(df["Name"])
print(df[["Name","Age"]])
print("========")
print(df.loc[0])
print(df.iloc[0,1])
print(df.iloc[1,2])
print("===")
print(df[df["Marks"]>40])

result = df[(df["Marks"]>40) & (df["Age"]>10)]
print(result)

result =df[(df["City"] == "Karachi") | (df["Marks"]> 60)]
print(result)

result = df[df["City"] .isin(["Karachi", "Lahore"])]
print(result)

result =df[df["Marks"].between(40, 60)]
print(result)

result = df.sort_values(by = "Marks")
print(result)

result = df.sort_values(by="Marks", ascending=False)
print(result)

result = df.sort_values(
    by =["City", "Marks"],
    ascending=[True, False]
)
print(result)

print("=====")
print(df.isnull)
print("=====")
print(df.isnull().sum())
print("======")
print(df.notnull())
```
